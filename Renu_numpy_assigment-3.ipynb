C2. Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy.

ans:
     The function standardizes each value using the z-score formula (x-mean)/std.

```python
import numpy as np

def standardize(x):
    mean = np.mean(x)
    std = np.std(x)
    return (x - mean) / std

x = np.array([10, 20, 30, 40, 50])
print("Original:",x)
print("Standardized:",standardize(x))
```

### D1. Data Preprocessing Pipeline 

You are given a raw 1D or 2D NumPy array containing NaN values.
Describe, step by step, the NumPy operations you would apply to:
(a) detect missing values,
(b) handle them appropriately and justify your strategy, and
(c) normalize or standardize the clean data before it is fed to a model.

ans:
     Data Preprocessing Pipeline:
                                 When working with real-world data, missing values can occur and are commonly represented by NaN in NumPy. Before feeding the data into a machine learning model, these missing values should be detected and handled properly so that they do not cause incorrect calculations or affect model training.

step-by-step:

Step 1 - Detect missing values:
                      
         I would use np.isnan() to identify which elements of the array contain NaN values. It returns a Boolean array where True indicates a missing value and False indicates a valid value. I can also use np.isnan(data).sum() to find the total number of missing values.

Step 2 - Handle missing values:

         After detecting the missing values, I would replace them with appropriate values instead of directly passing them to the model. For numerical data, replacing a missing value with the mean of its feature is one possible strategy. I would use np.nanmean() to calculate the mean while ignoring existing NaN values. This approach allows us to retain the available samples instead of removing complete rows.

For this example, I will use column-wise mean imputation because the data is 2D and each column represents a numerical feature. This keeps the replacement value specific to each feature.

Step 3 - Verify the cleaned data:

         After replacing the NaN values, I would check the array again using np.isnan() to make sure that no missing values remain. This is important before performing further preprocessing.

Step 4 - Standardize the clean data:

        Once the missing values have been handled, I would standardize the numerical features. Standardization subtracts the mean of each feature and divides by its standard deviation. The resulting features have approximately mean 0 and standard deviation 1. This can help machine learning models when the features have different scales.

The standardization formula is:

          standardized_data = (data - mean) / standard_deviation

I would calculate the mean and standard deviation column-wise using axis=0. NumPy broadcasting then applies these values to every row without requiring an explicit Python loop.

```python
data = np.array([ [10.0, 20.0], 
    [15.0, np.nan], 
    [20.0, 40.0], 
    [25.0, 50.0]
]) 

print("Missing values:") 
print(np.isnan(data)) 

column_mean = np.nanmean(data, axis=0) 

clean_data = np.where( np.isnan(data), column_mean, data ) 
 
mean = clean_data.mean(axis=0) 
std = clean_data.std(axis=0) 

standardized_data = (clean_data - mean) / std 
print("Clean data:") 
print(clean_data)
print("Standardized data:") 
print(standardized_data)
```

The original expression is:

Because exponentiation has higher precedence than subtraction, Python evaluates this as:

np.mean(y_true - (y_pred ** 2))

For Mean Squared Error, the difference must be calculated first and then squared. Therefore, the correct expression is:

np.mean((y_true - y_pred) ** 2)

```python
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])

b = 2.0

y_pred = X @ w + b

y_true = np.array([10, 20, 30])

mse = np.mean((y_true - y_pred) ** 2)

print("y_pred =", y_pred)
print("MSE =", mse)
```

E1. Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.

ans:

Min-max normalization scales each feature to a range between 0 and 1. Since this question requires per-column normalization, the minimum and maximum values are calculated separately for each feature column using axis=0.

The normalization formula is:

(X - column_min) / (column_max - column_min)

NumPy broadcasting automatically applies each column's minimum and maximum to the corresponding column. Therefore, the entire operation is vectorized and does not require explicit loops over rows or columns.

```python
X = np.array([
    [10, 100, 1000],
    [20, 200, 2000],
    [30, 300, 3000],
    [40, 400, 4000]
])
# Finding the minimum and maximum for each column
column_min = X.min(axis=0)
column_max = X.max(axis=0)
# Per-column min-max normalization
X_normalized = (X - column_min) / (column_max - column_min)
print("Original matrix:")
print(X)
print("\nColumn minimums:")
print(column_min)
print("\nColumn maximums:")
print(column_max)
print("\nNormalized matrix:")
print(X_normalized)
```

E2. Three-Level Score Classification 
Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each score as:
Fail (< 40), Average (40–70 inclusive), or Excellent (> 70).
Hint: You may need to nest or combine np.where() calls.

ans:
   The exam scores need to be classified into three categories based on their values. Scores below 40 are classified as "Fail", scores from 40 to 70 inclusive are classified as "Average", and scores above 70 are classified as "Excellent".

   I use nested np.where() to perform all three classifications in a single vectorized expression. The first np.where() checks whether a score is below 40. If it is not, the second np.where() checks whether the score is less than or equal to 70. This avoids explicit loops and if/else statements.

code explanetion:
          ->The outer np.where() checks whether each score is less than 40 and assigns "Fail" when the condition is true. For the remaining scores, 
          ->The inner np.where() checks whether the score is less than or equal to 70 and assigns "Average"; otherwise it assigns "Excellent".
          ->The entire classification is performed using one vectorized expression, so no explicit Python loops or if/else statements are required.

```python
import numpy as np 
scores = np.array([25, 40, 55, 70, 71, 90]) 
result = np.where(
     scores < 40,
      'Fail', 
np.where(scores <= 70, 'Average', 'Excellent') )
print(result)
```
